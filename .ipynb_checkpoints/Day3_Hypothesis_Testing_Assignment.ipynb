## Question 11 – Multiple Comparisons

### Objective
Simulate 5 groups of 30 values from the same normal distribution.
Perform all 10 pairwise t-tests and repeat the experiment 2,000 times.
Calculate the fraction of experiments with at least one significant result.

```python
# Set random seed
rng = np.random.default_rng(42)

# Create 5 groups
groups = []

for i in range(5):
    data = rng.normal(50, 10, 30)
    groups.append(data)

print("Number of groups:", len(groups))

# Display group means
for i, group in enumerate(groups):
    print(f"Group {i+1} mean:", round(np.mean(group), 2))
```

### Step 3 – Pairwise T-Tests

Compare the means of all possible pairs of groups using independent two-sample t-tests.

```python
# Generate all possible pairs
pairs = list(combinations(range(5), 2))

print("Total comparisons:", len(pairs))

# Perform t-tests
for a, b in pairs:

    t_stat, p_value = stats.ttest_ind(
        groups[a],
        groups[b]
    )

    print(
        f"Group {a+1} vs Group {b+1}: "
        f"t-statistic = {t_stat:.4f}, "
        f"p-value = {p_value:.4f}"
    )
```

```python
# Generate all possible pairs
pairs = list(combinations(range(5), 2))

print("Total comparisons:", len(pairs))

# Perform t-tests
for a, b in pairs:

    t_stat, p_value = stats.ttest_ind(
        groups[a],
        groups[b]
    )

    print(
        f"Group {a+1} vs Group {b+1}: "
        f"t-statistic = {t_stat:.4f}, "
        f"p-value = {p_value:.4f}"
    )
```

### Step 4 – Multiple Comparison Simulation

Repeat the experiment 2,000 times. In each experiment, generate five groups and perform all 10 pairwise t-tests. Count experiments where at least one p-value is below 0.05.

```python
# Number of simulations
simulations = 2000

# Count experiments with at least one false positive
false_positives = 0

for simulation in range(simulations):

    # Generate 5 groups
    groups = [
        rng.normal(50, 10, 30)
        for _ in range(5)
    ]

    # Store p-values
    p_values = []

    # Perform all 10 pairwise t-tests
    for a, b in combinations(range(5), 2):

        t_stat, p_value = stats.ttest_ind(
            groups[a],
            groups[b]
        )

        p_values.append(p_value)

    # Check whether any comparison is significant
    if any(p < 0.05 for p in p_values):
        false_positives += 1

print("Total simulations:", simulations)

print(
    "Experiments with at least one significant result:",
    false_positives
)

false_positive_rate = false_positives / simulations

print(
    "False positive rate:",
    round(false_positive_rate * 100, 2),
    "%"
)
```

```python
theoretical_rate = 1 - (0.95 ** 10)

print(
    "Theoretical probability:",
    round(theoretical_rate * 100, 2),
    "%"
)

print(
    "Simulated probability:",
    round(false_positive_rate * 100, 2),
    "%"
)
```

### Conclusion

The simulation was repeated 2,000 times using five groups
of 30 observations each.

The simulated false positive rate was 27.5%, while the
theoretical probability under the independence assumption
was 40.13%.

This demonstrates that performing multiple statistical
tests increases the chance of obtaining at least one
false positive.

Therefore, multiple comparison correction methods such
as Bonferroni correction are useful.

## Question 12 – Sample Size Calculation

Calculate the number of users required in each group
to detect an increase in conversion rate from 11.2%
to 12.2%, with 80% statistical power.

```python
from scipy.stats import norm
import math

# Given values
p1 = 0.112
p2 = 0.122

alpha = 0.05
power = 0.80

# Calculate average conversion rate
p_avg = (p1 + p2) / 2

# Critical Z values
z_alpha = norm.ppf(1 - alpha / 2)
z_power = norm.ppf(power)

# Calculate sample size
numerator = (
    z_alpha * math.sqrt(2 * p_avg * (1 - p_avg))
    +
    z_power * math.sqrt(
        p1 * (1 - p1) + p2 * (1 - p2)
    )
) ** 2

sample_size = numerator / (p2 - p1) ** 2

required_users = math.ceil(sample_size)

print("Required sample size per group:", required_users)
print("Total required users:", required_users * 2)
```

```python
n_a = 9800
n_b = 10050

print("Group A users:", n_a)
print("Group B users:", n_b)

print("Required per group:", required_users)

if n_a >= required_users and n_b >= required_users:
    print("Experiment has enough users.")
else:
    print("Experiment does not have enough users.")
```

### Conclusion

The experiment requires approximately 16,217 users
in each group to achieve 80% statistical power at a
5% significance level.

However, Group A contains 9,800 users and Group B
contains 10,050 users.

Therefore, the experiment does not have the required
sample size to achieve the planned statistical power.
The company should collect more data before making
a final decision.

## Question 13 – Sample Ratio Mismatch (SRM)

Check whether users are distributed equally between
Group A and Group B using a Chi-square goodness-of-fit test.

Expected distribution = 50% A and 50% B.
Significance level = 0.05.

```python
# Observed user counts
observed = np.array([n_a, n_b])

# Total users
total_users = observed.sum()

# Expected 50/50 distribution
expected = np.array([
    total_users * 0.5,
    total_users * 0.5
])

print("Observed counts:", observed)
print("Expected counts:", expected)
```

```python
chi2_stat, p_value = stats.chisquare(
    f_obs=observed,
    f_exp=expected
)

print("Chi-square statistic:", round(chi2_stat, 4))
print("P-value:", round(p_value, 4))

alpha = 0.05

if p_value < alpha:
    print("Sample Ratio Mismatch detected!")
else:
    print("No significant Sample Ratio Mismatch detected.")
```

### Conclusion

The Chi-square goodness-of-fit test was performed
to check whether users were distributed equally
between Group A and Group B.

The observed counts were 9,800 and 10,050,
while the expected counts were 9,925 in each group.

The Chi-square statistic was approximately 3.1486
and the p-value was approximately 0.0760.

Since the p-value is greater than 0.05, we fail to
reject the null hypothesis.

Therefore, no statistically significant sample
ratio mismatch was detected.

## Question 14 – Two-Proportion Z-Test

Compare conversion rates between Group A and Group B.

Calculate:
- Conversion rate for both groups
- Absolute lift
- Relative lift
- Z-statistic and p-value
- 95% confidence interval for the difference

```python
# Generate the assignment dataset
rng = np.random.default_rng(2026)

n_a, n_b = 9800, 10050

ab = pd.DataFrame({
    "group": ["A"] * n_a + ["B"] * n_b,

    "converted": np.r_[
        rng.random(n_a) < 0.112,
        rng.random(n_b) < 0.121
    ].astype(int),

    "order_value": np.r_[
        rng.gamma(4, 300, n_a),
        rng.gamma(4, 305, n_b)
    ]
})

ab.loc[ab.converted == 0, "order_value"] = 0

print("Dataset created successfully!")
print(ab.head())
print(ab.shape)
```

```python
# Count conversions in each group
conversions = ab.groupby("group")["converted"].sum()

# Count users in each group
users = ab.groupby("group")["converted"].count()

# Calculate conversion rates
rate_a = conversions["A"] / users["A"]
rate_b = conversions["B"] / users["B"]

print("Group A conversions:", conversions["A"])
print("Group B conversions:", conversions["B"])

print("Group A conversion rate:", round(rate_a * 100, 2), "%")
print("Group B conversion rate:", round(rate_b * 100, 2), "%")
```

```python
# Absolute lift
absolute_lift = rate_b - rate_a

# Relative lift
relative_lift = (absolute_lift / rate_a) * 100

print("Absolute lift:", round(absolute_lift * 100, 2), "percentage points")

print("Relative lift:", round(relative_lift, 2), "%")
```

```python
from scipy.stats import norm

# Number of conversions
x_a = conversions["A"]
x_b = conversions["B"]

# Total users
n_a = users["A"]
n_b = users["B"]

# Pooled conversion rate
pooled_rate = (x_a + x_b) / (n_a + n_b)

# Standard error under the null hypothesis
se = np.sqrt(
    pooled_rate * (1 - pooled_rate) *
    (1/n_a + 1/n_b)
)

# Z-statistic
z_stat = (rate_b - rate_a) / se

# Two-sided p-value
p_value = 2 * norm.sf(abs(z_stat))

print("Z-statistic:", round(z_stat, 4))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Statistically significant difference.")
else:
    print("No statistically significant difference.")
```

```python
# Standard error for confidence interval
se_ci = np.sqrt(
    rate_a * (1-rate_a) / n_a +
    rate_b * (1-rate_b) / n_b
)

# Difference in conversion rates
difference = rate_b - rate_a

# 95% confidence interval
lower = difference - 1.96 * se_ci
upper = difference + 1.96 * se_ci

print("95% Confidence Interval:")
print("Lower limit:", round(lower * 100, 2), "%")
print("Upper limit:", round(upper * 100, 2), "%")
```

### Conclusion

A two-proportion Z-test was performed to compare
conversion rates between Group A and Group B.

Group A had a conversion rate of approximately 10.77%,
while Group B had a conversion rate of approximately
11.79%.

The absolute lift was approximately 1.03 percentage
points, and the relative lift was approximately 9.53%.

The Z-statistic was approximately 2.2835 and the
p-value was approximately 0.0224.

Since the p-value is less than 0.05, we reject the
null hypothesis.

Therefore, there is a statistically significant
difference in conversion rates between the two groups.
Group B performed better than Group A in this experiment.

## Question 15 – Average Order Value Comparison

Compare order values between converted users in
Group A and Group B.

Since order values are skewed, use the
Mann-Whitney U test.

```python
# Select only converted users
converted_users = ab[ab["converted"] == 1]

# Separate order values
orders_a = converted_users[
    converted_users["group"] == "A"
]["order_value"]

orders_b = converted_users[
    converted_users["group"] == "B"
]["order_value"]

print("Group A converted users:", len(orders_a))
print("Group B converted users:", len(orders_b))
```

```python
print("Group A average order value:", round(orders_a.mean(), 2))

print("Group B average order value:", round(orders_b.mean(), 2))

print("Group A median:", round(orders_a.median(), 2))

print("Group B median:", round(orders_b.median(), 2))
```

```python
# Perform Mann-Whitney U test
u_stat, p_value = stats.mannwhitneyu(
    orders_a,
    orders_b,
    alternative="two-sided"
)

print("Mann-Whitney U statistic:", round(u_stat, 2))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("There is a statistically significant difference.")
else:
    print("There is no statistically significant difference.")
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=converted_users,
    x="group",
    y="order_value"
)

plt.title("Order Value Comparison: Group A vs Group B")
plt.xlabel("Group")
plt.ylabel("Order Value")

plt.show()
```

```python
print("GROUP A RESULTS")
print("Number of orders:", len(orders_a))
print("Mean:", round(orders_a.mean(), 2))
print("Median:", round(orders_a.median(), 2))

print("\nGROUP B RESULTS")
print("Number of orders:", len(orders_b))
print("Mean:", round(orders_b.mean(), 2))
print("Median:", round(orders_b.median(), 2))

print("\nMANN-WHITNEY U TEST")
print("U statistic:", round(u_stat, 2))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Significant difference between groups.")
else:
    print("No significant difference between groups.")
```

### Conclusion

Only converted users were selected for the analysis.

The average order value was 1208.40 for Group A
and 1230.18 for Group B.

Since order values were skewed, the Mann-Whitney U
test was used.

The U statistic was 617346 and the p-value was 0.6124.

Since the p-value is greater than 0.05, we fail to
reject the null hypothesis.

Therefore, there is no statistically significant
difference in order-value distributions between
Group A and Group B.

## Question 16 – Peeking Simulation

Simulate an A/A test where both groups have a true
conversion rate of 11.2%.

Check the p-value after every 500 users per group,
up to 10,000 users.

Stop when the first p-value is below 0.05.

Repeat the experiment 1,000 times and calculate
the false-positive rate.

```python
# Question 16: Peeking Simulation

from scipy.stats import norm

# Given values
true_rate = 0.112
simulations = 1000
max_users = 10000
check_interval = 500
alpha = 0.05

# Random number generator
rng_peek = np.random.default_rng(2026)

# Count false positives
false_positives = 0

# Store stopping points
stopping_points = []

for experiment in range(simulations):

    # Generate A/A test data
    group_a = rng_peek.binomial(1, true_rate, max_users)
    group_b = rng_peek.binomial(1, true_rate, max_users)

    significant_found = False

    # Check after every 500 users
    for n in range(check_interval, max_users + 1, check_interval):

        # Count conversions
        x_a = group_a[:n].sum()
        x_b = group_b[:n].sum()

        # Calculate conversion rates
        rate_a = x_a / n
        rate_b = x_b / n

        # Pooled conversion rate
        pooled = (x_a + x_b) / (2 * n)

        # Standard error
        se = np.sqrt(
            pooled * (1 - pooled) * (2 / n)
        )

        # Z-statistic
        if se == 0:
            continue

        z_stat = (rate_a - rate_b) / se

        # Two-sided p-value
        p_value = 2 * norm.sf(abs(z_stat))

        # Stop at first significant result
        if p_value < alpha:

            false_positives += 1
            stopping_points.append(n)

            significant_found = True
            break

    # No significant result by 10,000
    if not significant_found:
        stopping_points.append(max_users)

# Calculate false-positive rate
false_positive_rate_peeking = (
    false_positives / simulations
)

print("Total experiments:", simulations)

print("False positive experiments:", false_positives)

print(
    "False positive rate:",
    round(false_positive_rate_peeking * 100, 2),
    "%"
)
```

```python
plt.figure(figsize=(7, 4))

plt.bar(
    ["Normal 5% threshold", "Peeking simulation"],
    [5, false_positive_rate_peeking * 100]
)

plt.ylabel("False Positive Rate (%)")
plt.title("Effect of Repeated Peeking")

plt.show()
```

### Conclusion

An A/A test was simulated 1,000 times with both
groups having a true conversion rate of 11.2%.

The p-value was checked after every 500 users per
group, up to 10,000 users.

The experiment stopped whenever the p-value became
less than 0.05.

Repeatedly checking the p-value increases the chance
of obtaining a false positive result.

This is called the peeking problem.

Therefore, checking results repeatedly without
proper statistical correction is dangerous.

A fixed sample size or sequential testing method
should be used to control the false-positive rate.
